```python
import os
import pandas as pd

# Define the path to the folder containing the files
folder_path = r'\\......\Tong_computer\Dissertation Study 2\EEG data\power_time_series'

# Create a mapping of channel numbers to EEG labels
channel_mapping = {
    'Channel 1': 'AF3',
    'Channel 2': 'F7',
    'Channel 3': 'F3',
    'Channel 4': 'FC5',
    'Channel 5': 'T7',
    'Channel 6': 'P7',
    'Channel 7': 'O1',
    'Channel 8': 'O2',
    'Channel 9': 'P8',
    'Channel 10': 'T8',
    'Channel 11': 'FC6',
    'Channel 12': 'F4',
    'Channel 13': 'F8',
    'Channel 14': 'AF4'
}

# Loop through all the files in the folder
for filename in os.listdir(folder_path):
    if filename.endswith('.xlsx'):  # Look for Excel files
        file_path = os.path.join(folder_path, filename)
        
        # Read the Excel file
        df = pd.read_excel(file_path)
        
        # Print original column names for debugging
        print(f"Original columns in {filename}:")
        print(df.columns)

        # Try replacing the channel numbers with EEG labels
        new_columns = []
        for col in df.columns:
            parts = col.split(' - ')  # Split on ' - ', assuming the format is 'Channel X - ...'
            if parts[0] in channel_mapping:
                new_column = channel_mapping[parts[0]] + ' - ' + parts[1]
                new_columns.append(new_column)
            else:
                new_columns.append(col)  # Keep the original column if no mapping is found
        
        # Assign new column names to the DataFrame
        df.columns = new_columns
        
        # Print new column names for debugging
        print(f"Updated columns in {filename}:")
        print(df.columns)
        
        # Save the updated file back to the same folder, overwriting the original file
        df.to_excel(file_path, index=False)

print("Channel numbers replacement process completed.")
```

```python
import os
import pandas as pd

# Define file paths
eeg_folder_path = r'\\......\Tong_computer\Dissertation Study 2\EEG data\power_time_series'
cumulative_file_path = r'\\......\Tong_computer\Dissertation Study 2\Study2 Raw Data\resampled_ratings\combined_file\combined_file_sad.xlsx'

# Load the cumulative file
cumulative_df = pd.read_excel(cumulative_file_path)

# Loop through all files in the EEG folder
for filename in os.listdir(eeg_folder_path):
    if filename.endswith('.xlsx') and not filename.startswith('~$'):  # Skip temporary files
        try:
            # Extract participant number from the filename (first 2 digits)
            participant_number = int(filename[:2])
        except ValueError:
            print(f"Skipping file {filename} as it does not contain a valid participant number.")
            continue
        
        # Load the individual participant EEG file
        eeg_file_path = os.path.join(eeg_folder_path, filename)
        eeg_df = pd.read_excel(eeg_file_path)
        
        # Check the format of the EEG file (ensure it has at least 60 rows)
        if len(eeg_df) != 60:
            print(f"Warning: EEG file {filename} does not have exactly 60 rows. Skipping this file.")
            continue
        
        # Find the rows in the cumulative file that correspond to this participant's first 60 seconds
        participant_rows = cumulative_df[(cumulative_df['participant'] == participant_number) & (cumulative_df['Video Time'].str.startswith('00:00:'))]
        
        # Ensure that there are exactly 60 matching rows in the cumulative file
        if len(participant_rows) != 60:
            print(f"Warning: Cumulative file does not have exactly 60 rows for participant {participant_number}. Skipping this participant.")
            continue
        
        # Get the indices of the corresponding rows in the cumulative file
        indices = participant_rows.index
        
        # Insert the EEG data into the corresponding rows in the cumulative file
        for i, row_index in enumerate(indices):
            # Assuming EEG data has consistent columns and they need to be appended to cumulative_df
            eeg_data = eeg_df.iloc[i].values
            cumulative_df.loc[row_index, eeg_df.columns] = eeg_data

# Save the updated cumulative file
cumulative_df.to_excel(cumulative_file_path, index=False)

print("EEG data has been successfully inserted into the cumulative file.")
```
