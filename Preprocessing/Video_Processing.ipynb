```python
import os
import subprocess

# Full path to ffmpeg executable
ffmpeg_path = r"C:\ffmpeg\bin"
ffmpeg_executable = r"C:\ffmpeg\bin\ffmpeg.exe"

# Update PATH environment variable
os.environ["PATH"] += os.pathsep + ffmpeg_path

# Directory containing the videos
input_directory = r"C:\Users\OLten\Desktop\Gutsell lab\empathy_video_psychopy_july4\original_videos"
output_directory = r"C:\Users\OLten\Desktop\Gutsell lab\empathy_video_psychopy_july4\videos"

# Create output directory if it doesn't exist
os.makedirs(output_directory, exist_ok=True)

# List of video file extensions to look for
video_extensions = ('.mp4', '.mov', '.avi', '.mkv', '.wmv')

# Loop through each file in the input directory and print video files
print("Video files found:")
for filename in os.listdir(input_directory):
    if filename.endswith(video_extensions):
        print(filename)

```

```python
# Loop through each file in the input directory and process video files
for filename in os.listdir(input_directory):
    if filename.endswith(video_extensions):
        input_path = os.path.join(input_directory, filename)
        output_path = os.path.join(output_directory, f"{os.path.splitext(filename)[0]}_extended{os.path.splitext(filename)[1]}")

        print(f"Processing file: {input_path}")
        
        # Command to extend the last frame by 2 seconds
        command = [
            ffmpeg_executable,
            '-i', input_path,
            '-vf', 'tpad=stop_mode=clone:stop_duration=2',
            '-c:a', 'copy',
            output_path
        ]
        
        try:
            # Capture stderr to debug ffmpeg errors
            result = subprocess.run(command, check=True, stderr=subprocess.PIPE, text=True)
            print(f"Successfully processed {input_path}")
        except subprocess.CalledProcessError as e:
            print(f"Error processing {input_path}: {e.stderr}")

print("Processing complete!")
```
