```python
import pandas as pd
import os

# Define the input folders
folder_1 = r'\\files.brandeis.edu\gutsell-lab\Tong_computer\Dissertation Study 2\Study2 Raw Data\resampled_ratings\facial_expression_happy_rating'
folder_2 = r'\\files.brandeis.edu\gutsell-lab\Tong_computer\Dissertation Study 2\Study2 Raw Data\resampled_ratings\facial_expression_sad_movie'
folder_3 = r'\\files.brandeis.edu\gutsell-lab\Tong_computer\Dissertation Study 2\Study2 Raw Data\resampled_ratings\happy_movie_rating'
folder_4 = r'\\files.brandeis.edu\gutsell-lab\Tong_computer\Dissertation Study 2\Study2 Raw Data\resampled_ratings\sad_movie_rating'

# Output folder
output_folder = r'\\files.brandeis.edu\gutsell-lab\Tong_computer\Dissertation Study 2\Study2 Raw Data\resampled_ratings\combined_file'

# Helper function to load files based on extension
def load_file(filepath):
    if filepath.endswith('.xlsx'):
        return pd.read_excel(filepath)
    elif filepath.endswith('.csv'):
        return pd.read_csv(filepath)
    else:
        raise ValueError("Unsupported file format: " + filepath)

# Loop through participants 01 to 22, skipping participant 15
for participant_num in range(1, 23):
    if participant_num == 15:
        continue  # Skip participant 15

    participant_str = f"{participant_num:02}"  # Format as two digits, e.g., '01', '02', etc.

    # Step 1: Find the file that begins with the participant number in folder 1
    file_1 = [f for f in os.listdir(folder_1) if f.startswith(participant_str)][0]
    data_1 = load_file(os.path.join(folder_1, file_1))

    # Step 2: Find the file that ends with the participant number in folder 3
    file_2 = [f for f in os.listdir(folder_3) if f.endswith(f"{participant_str}.xlsx") or f.endswith(f"{participant_str}.csv")][0]
    data_2 = load_file(os.path.join(folder_3, file_2))

    # Step 3: Combine data_2 to the right of data_1
    combined_1 = pd.concat([data_1, data_2], axis=1)

    # Step 4: Find the file that begins with the participant number in folder 2
    file_3 = [f for f in os.listdir(folder_2) if f.startswith(participant_str)][0]
    data_3 = load_file(os.path.join(folder_2, file_3))

    # Step 5: Find the file that ends with the participant number in folder 4
    file_4 = [f for f in os.listdir(folder_4) if f.endswith(f"{participant_str}.xlsx") or f.endswith(f"{participant_str}.csv")][0]
    data_4 = load_file(os.path.join(folder_4, file_4))

    # Step 6: Combine data_4 to the right of data_3
    combined_2 = pd.concat([data_3, data_4], axis=1)

    # Step 7: Add "condition" column
    combined_1["condition"] = "positive"
    combined_2["condition"] = "negative"

    # Step 8: Append combined_2 to combined_1
    final_combined = pd.concat([combined_1, combined_2])

    # Step 9: Insert a new column "participant" with the participant number
    final_combined.insert(0, "participant", participant_str)

    # Step 10: Save the final combined file
    output_file_path = os.path.join(output_folder, f'combined_{participant_str}.xlsx')
    final_combined.to_excel(output_file_path, index=False)

    print(f"Combined file for participant {participant_str} saved to {output_file_path}")
```

```python
#########################  combine all files
import pandas as pd
import os

# Define the folder containing the files
input_folder = r'\\files.brandeis.edu\gutsell-lab\Tong_computer\Dissertation Study 2\Study2 Raw Data\resampled_ratings\combined_file'

# Output file path
output_file_path = os.path.join(input_folder, 'combined_file.xlsx')

# Initialize an empty list to store the dataframes
all_data = []

# Loop through all the files in the folder
for filename in os.listdir(input_folder):
    if filename.endswith('.xlsx'):  # Process only .xlsx files
        file_path = os.path.join(input_folder, filename)
        
        # Read each Excel file and append to the list
        df = pd.read_excel(file_path)
        all_data.append(df)

# Concatenate all dataframes
combined_df = pd.concat(all_data, ignore_index=True)

# Save the combined dataframe to a new Excel file
combined_df.to_excel(output_file_path, index=False)

print(f"All files combined and saved to {output_file_path}")
```
