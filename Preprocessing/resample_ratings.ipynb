```python
import pandas as pd
import os

# File paths
input_folder = r'\......\Tong_computer\Dissertation Study 2\Study2 Raw Data'
output_folder = r'\......\Tong_computer\Dissertation Study 2\Study2 Raw Data\resampled_ratings'

# Create the output folder if it doesn't exist
if not os.path.exists(output_folder):
    os.makedirs(output_folder)

# Loop through all files in the input folder
for filename in os.listdir(input_folder):
    # Only process files that contain 'rating' in the filename
    if 'rating' in filename and filename.endswith('.csv'):
        input_file_path = os.path.join(input_folder, filename)
        
        # Load the file
        data = pd.read_csv(input_file_path)
        
        # Rename columns for clarity (assuming first column is ratings and second is timestamps)
        data.columns = ['emotional_rating', 'timestamp_seconds']

        # Round the timestamp to the nearest second
        data['rounded_time'] = data['timestamp_seconds'].round()

        # Group by the rounded time and calculate the mean emotional rating for each second
        data_resampled = data.groupby('rounded_time', as_index=False).mean()

        # Filter the data to keep only the first 180 seconds (3 minutes)
        data_resampled = data_resampled[(data_resampled['rounded_time'] >= 0) & (data_resampled['rounded_time'] <= 180)]

        # Rename columns to match the desired format
        data_resampled = data_resampled[['rounded_time', 'emotional_rating']].rename(columns={'rounded_time': 'timestamp_seconds'})

        # Save the result to the output folder with a new filename
        output_file_path = os.path.join(output_folder, f'resampled_{filename}')
        data_resampled.to_csv(output_file_path, index=False)

        print(f"Resampled file saved to {output_file_path}")
```

```python
import pandas as pd
import os

# Input and output folder paths
input_folder = r'E:\Tong_empathy_video_study\empathy_video_study'
output_folder = r'\......\Tong_computer\Dissertation Study 2\Study2 Raw Data\resampled_ratings\facial_expression_sad_movie'

# Replace column names as per the screenshot
new_column_names = ['Video Time', 'Neutral', 'Happy', 'Sad', 'Angry', 'Surprised', 'Scared', 'Disgusted', 'Valence', 'Arousal']

# Ensure output folder exists
if not os.path.exists(output_folder):
    os.makedirs(output_folder)

# Process all files that contain "longVideo" in their name
for filename in os.listdir(input_folder):
    if "longVideo" in filename and filename.endswith(".xlsx"):
        input_file_path = os.path.join(input_folder, filename)
        output_file_name = filename  # Using the same name for output
        output_file_path = os.path.join(output_folder, output_file_name)

        # Step 1: Load the Excel file
        data = pd.read_excel(input_file_path)

        # Step 2: Find the row index where the first column is "Video Time"
        start_index = data[data.iloc[:, 0] == "Video Time"].index[0]

        # Step 3: Find the row index where the first column is "00:02:59.000"
        end_index = data[data.iloc[:, 0] == "00:02:59.000"].index[0]

        # Step 4: Filter the data from the "Video Time" row to the "00:02:59.000" row
        filtered_data = data.iloc[start_index:end_index + 1]

        # Step 5: Save the filtered data to a new Excel file
        filtered_data.to_excel(output_file_path, index=False)

        # Step 6: Now reload the saved file and modify the header
        data = pd.read_excel(output_file_path)

        # Step 7: Drop the first row (old headers and metadata)
        data = data.drop(index=0).reset_index(drop=True)

        # Step 8: Replace the column names with the correct ones from the screenshot
        data.columns = new_column_names

        # Step 9: Save the updated file with the new headers
        data.to_excel(output_file_path, index=False)

        print(f"Processed and saved: {output_file_path}")
```
