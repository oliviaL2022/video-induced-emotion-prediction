```python
############## Making the first intial combined power and entropy files #####################
import pandas as pd
import os

def get_matching_files(power_dir, entropy_dir):
    power_files = {f[:2]: os.path.join(power_dir, f) for f in os.listdir(power_dir) if f.endswith('.xlsx')}
    entropy_files = {f[:2]: os.path.join(entropy_dir, f) for f in os.listdir(entropy_dir) if f.endswith('.xlsx')}
    
    # Find common prefixes
    common_keys = set(power_files.keys()).intersection(set(entropy_files.keys()))
    
    return [(power_files[k], entropy_files[k]) for k in common_keys]

def merge_and_save(power_file, entropy_file, save_dir):
    power_df = pd.read_excel(power_file)
    entropy_df = pd.read_excel(entropy_file)
    
    # Ensure time column consistency
    power_df.rename(columns={power_df.columns[0]: 'Time (s)'}, inplace=True)
    entropy_df.rename(columns={entropy_df.columns[0]: 'Time (s)'}, inplace=True)
    
    # Merge on 'Time (s)'
    combined_df = pd.merge(power_df, entropy_df, on='Time (s)', how='inner')
    
    # Generate save path
    filename = os.path.basename(power_file).replace('_power_time_series_new.xlsx', '_combined.csv')
    save_path = os.path.join(save_dir, filename)
    
    combined_df.to_csv(save_path, index=False)
    print(f"Combined file saved to: {save_path}")

# Define directories
power_dir = r"\\files.brandeis.edu\gutsell-lab\Tong_computer\Dissertation Study 2\EEG data\power_time_series\q1"
entropy_dir = r"\\files.brandeis.edu\gutsell-lab\Tong_computer\Dissertation Study 2\EEG data\entropy\q1"
save_dir = r"\\files.brandeis.edu\gutsell-lab\Tong_computer\Dissertation Study 2\EEG data\Combined Files\combined_power_entropy"

# Ensure save directory exists
os.makedirs(save_dir, exist_ok=True)

# Process matching files
matching_files = get_matching_files(power_dir, entropy_dir)
for power_file, entropy_file in matching_files:
    merge_and_save(power_file, entropy_file, save_dir)
```

```python
################################## Adding the first video, the practice video ########################
import pandas as pd
import os
import re

def extract_subject_id(filename):
    numbers = re.findall(r'\d+', filename)  # Extract all numbers from the filename
    if numbers:
        subject_id = numbers[0]  # Take the first number found
        return subject_id.zfill(2)  # Ensure it's a two-digit number
    return None

def get_matching_files(eeg_dir, video_dir):
    eeg_files = {f[:2]: os.path.join(eeg_dir, f) for f in os.listdir(eeg_dir) if f.endswith('.csv')}
    video_files = {}
    
    for f in os.listdir(video_dir):
        if f.endswith('.xlsx'):
            subject_id = extract_subject_id(f)
            if subject_id:
                video_files[subject_id] = os.path.join(video_dir, f)
    
    print(f"Number of EEG files found: {len(eeg_files)}")
    print(f"Number of video files found: {len(video_files)}")
    
    # Find common subject numbers
    common_keys = set(eeg_files.keys()).intersection(set(video_files.keys()))
    
    return [(eeg_files[k], video_files[k]) for k in common_keys]

def merge_and_save(eeg_file, video_file, save_dir):
    eeg_df = pd.read_csv(eeg_file)
    video_df = pd.read_excel(video_file, header=None)  # Read without a header
    
    # Start with the second row of the entire data
    column_names = video_df.iloc[0].tolist()
    video_data = video_df.iloc[1:].dropna(how="all").reset_index(drop=True).iloc[:60]
    video_data.columns = column_names
    
    # Check if `video_data` has fewer than 60 rows
    # It pads missing rows with pd.NA so the code won't crash.
    if video_data.shape[0] < 60:
        missing_rows = 60 - video_data.shape[0]
        print(f"WARNING: Video file {video_file} has only {video_data.shape[0]} rows instead of 60. Padding {missing_rows} missing rows.")
        padding_df = pd.DataFrame([[pd.NA] * video_data.shape[1]] * missing_rows, columns=video_data.columns)
        video_data = pd.concat([video_data, padding_df], ignore_index=True)
        
    # Find the row index in EEG data where "Channel 1 - Emotion_Condition" is "long_happy"
    matching_rows = eeg_df[eeg_df["Channel 1 - Emotion_Condition"] == "long_happy"].index
    
    if len(matching_rows) > 0:
        insert_idx = matching_rows[0]
        
        # Ensure EEG DataFrame has space for new columns
        for col in column_names:
            if col not in eeg_df.columns:
                eeg_df[col] = pd.NA
        
        # Append only 60 rows of video data to EEG file at the identified index
        eeg_df.iloc[insert_idx:insert_idx+60, eeg_df.columns.get_loc(column_names[0]):] = video_data.values
        
        # Generate save path
        filename = os.path.basename(eeg_file).replace(".csv", "_updated1.csv")
        save_path = os.path.join(save_dir, filename)
        
        eeg_df.to_csv(save_path, index=False)
        print(f"Updated EEG file saved as: {save_path}")
    else:
        print("No 'long_happy' condition found in EEG file.")

# Define directories
eeg_dir = r"\\files.brandeis.edu\gutsell-lab\Tong_computer\Dissertation Study 2\EEG data\Combined Files\combined_power_entropy"
video_dir = r"\\files.brandeis.edu\gutsell-lab\Tong_computer\Dissertation Study 2\Video Data\practice video\FaceReader Output\corrected practice video"
save_dir = r"\\files.brandeis.edu\gutsell-lab\Tong_computer\Dissertation Study 2\EEG data\Combined Files\power_entropy_updated1"

# Ensure save directory exists
os.makedirs(save_dir, exist_ok=True)

# Process matching files
matching_files = get_matching_files(eeg_dir, video_dir)
for eeg_file, video_file in matching_files:
    merge_and_save(eeg_file, video_file, save_dir)
```

```python
############## update by adding the unsegmented video 
import pandas as pd
import os
import re

def extract_subject_id(filename):
    numbers = re.findall(r'\d+', filename)  # Extract all numbers from the filename
    if numbers:
        subject_id = numbers[0]  # Take the first number found
        return subject_id.zfill(2)  # Ensure it's a two-digit number
    return None

def get_matching_files(base_dir, video_dir):
    base_files = {f[:2]: os.path.join(base_dir, f) for f in os.listdir(base_dir) if f.endswith('.csv')}
    video_files = {}
    
    for f in os.listdir(video_dir):
        if f.endswith('.xlsx'):
            subject_id = extract_subject_id(f)
            if subject_id:
                video_files[subject_id] = os.path.join(video_dir, f)
    
    print(f"Number of base files found: {len(base_files)}")
    print(f"Number of video files found: {len(video_files)}")
    
    # Find common subject numbers
    common_keys = set(base_files.keys()).intersection(set(video_files.keys()))
    
    return [(base_files[k], video_files[k]) for k in common_keys]

def merge_and_save(base_file, video_file, save_dir):
    df_base = pd.read_csv(base_file)
    df_video = pd.read_csv(video_file)
    
    print(f"Processing: {base_file} and {video_file}")
    print("Columns in the 'base_df' (EEG file):", df_base.columns.tolist())
    print("Columns in the 'video_df' (Video file):", df_video.columns.tolist())
    
    # Columns to append (from Neutral to Quality)
    columns_to_append = df_base.loc[:, 'Neutral':'Quality'].columns.tolist()
    
    # Additional columns from video_file to add
    df_video_extra_columns = ['stimID']
    
    # Ensure base file includes these extra columns
    df_base = df_base.assign(**{col: None for col in df_video_extra_columns})
    
    # Find where to insert chunks
    emotion_indices = df_base[df_base['Channel 1 - Emotion_Condition'].isin(['emotional', 'neutral'])].index.tolist()
    
    # Preserve original order of stimID chunks
    stim_chunks = [chunk for _, chunk in df_video.groupby((df_video['stimID'] != df_video['stimID'].shift()).cumsum())]
    
    # Make sure we have enough chunks to match the indices
    assert len(stim_chunks) >= len(emotion_indices), "Not enough chunks in video_file to match base_file occurrences."
    
    # Append data to base_file and expand Channel 1 - Emotion_Condition
    for i, start_idx in enumerate(emotion_indices):
        chunk_df = stim_chunks[i]  # Extract chunk data (in original order)
        end_idx = start_idx + len(chunk_df)  # Define end range
        
        # Ensure end_idx doesn't exceed the base file size
        end_idx = min(end_idx, len(df_base))
        
        # Fill the corresponding columns
        df_base.loc[start_idx:end_idx-1, columns_to_append] = chunk_df[columns_to_append].values
        df_base.loc[start_idx:end_idx-1, df_video_extra_columns] = chunk_df[df_video_extra_columns].values
        
        # Expand Channel 1 - Emotion_Condition to all rows in the chunk
        df_base.loc[start_idx:end_idx-1, 'Channel 1 - Emotion_Condition'] = df_base.loc[start_idx, 'Channel 1 - Emotion_Condition']
    
    # Save the modified file
    filename = os.path.basename(base_file).replace(".csv", "_combined.csv")
    save_path = os.path.join(save_dir, filename)
    df_base.to_csv(save_path, index=False)
    print(f"Updated file saved at: {save_path}")

# Define directories
base_dir = r"\\files.brandeis.edu\gutsell-lab\Tong_computer\Dissertation Study 2\EEG data\Combined Files\power_entropy_updated2"
video_dir = r"\\files.brandeis.edu\gutsell-lab\Tong_computer\Dissertation Study 2\Video Data\unsegmented videos\FaceReader Output\unsegmented video\updated_video_outputs"
save_dir = r"\\files.brandeis.edu\gutsell-lab\Tong_computer\Dissertation Study 2\EEG data\Combined Files\combined_power_entropy_video"

# Ensure save directory exists
os.makedirs(save_dir, exist_ok=True)

# Process matching files
matching_files = get_matching_files(base_dir, video_dir)
for base_file, video_file in matching_files:
    merge_and_save(base_file, video_file, save_dir)
```
