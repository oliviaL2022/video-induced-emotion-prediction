```python
import pandas as pd
import os
import glob
import re

# Define paths
video_output_path = r'\\......\Tong_computer\Dissertation Study 2\Video Data\unsegmented videos\FaceReader Output\unsegmented video'
psychopy_data_path = r'\\......\Tong_computer\Dissertation Study 2\Raw Data\Behavioral_psychopy'
new_output_folder = os.path.join(video_output_path, 'updated_video_outputs')
ttl_output_folder = r'\\......\Tong_computer\Dissertation Study 2\EEG data\ttl_from_set\new_ttl\video segmentation ttl'

# Create the new folders if they do not exist
os.makedirs(new_output_folder, exist_ok=True)
os.makedirs(ttl_output_folder, exist_ok=True)

# Get all video output files
video_files = glob.glob(os.path.join(video_output_path, '*.xlsx'))

for video_file in video_files:
    # Extract subject ID from filename using regex
    video_filename = os.path.basename(video_file)
    match = re.search(r'\d+', video_filename)  # Find the first number in the filename
    if not match:
        print(f'No subject ID found in filename: {video_filename}')
        continue
    
    subject_id = match.group().zfill(2)  # Ensure two-digit format
    
    # Find corresponding PsychoPy file
    psychopy_file = glob.glob(os.path.join(psychopy_data_path, f'{subject_id}*.csv'))
    
    if not psychopy_file:
        print(f'No matching PsychoPy file found for subject {subject_id}')
        continue
    
    psychopy_file = psychopy_file[0]  # Take the first match
    
    # Read the video output file and psychopy file
    video_df = pd.read_excel(video_file)
    psycho_df = pd.read_csv(psychopy_file)
    
    if 'stimID' not in psycho_df.columns or 'video_q1.started' not in psycho_df.columns or 'movie_clip1.started' not in psycho_df.columns:
        print(f'Skipping {psychopy_file} due to missing required columns.')
        continue
    
    # Extract stimID and its corresponding duration
    stim_id_list = []
    ttl_data = []
    
    for index, row in psycho_df.iterrows():
        stim_id = row['stimID']
        if pd.notna(stim_id):  # First non-empty stimID
            duration = round(row['video_q1.started'] - row['movie_clip1.started'])
            stim_id_list.extend([stim_id] * duration)
            ttl_data.append({'stimID': stim_id, 'duration': duration})
    
    # Trim or extend stim_id_list to match video_df length
    video_length = len(video_df)
    if len(stim_id_list) > video_length:
        stim_id_list = stim_id_list[:video_length]
    else:
        stim_id_list.extend([None] * (video_length - len(stim_id_list)))
    
    # Add stimID column to video_df
    video_df['stimID'] = stim_id_list
    
    # Save updated file in the new folder with proper formatting
    output_file = os.path.join(new_output_folder, f'updated_{video_filename}')
    video_df.to_csv(output_file, index=False, encoding='utf-8-sig')  # Ensure proper encoding for Excel
    
    # Save TTL data for segmentation
    ttl_df = pd.DataFrame(ttl_data)
    ttl_output_file = os.path.join(ttl_output_folder, f'ttl_{subject_id}.csv')
    ttl_df.to_csv(ttl_output_file, index=False, encoding='utf-8-sig')
    
    print(f'Updated file saved: {output_file}')
    print(f'TTL file saved: {ttl_output_file}')
```
